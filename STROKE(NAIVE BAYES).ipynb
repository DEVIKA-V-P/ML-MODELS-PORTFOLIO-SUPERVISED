```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings("ignore")
from sklearn.preprocessing import MinMaxScaler,LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,ConfusionMatrixDisplay
from sklearn.naive_bayes import GaussianNB,MultinomialNB,BernoulliNB
```

```python
df=pd.read_csv("13.stroke.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df=df.drop(['id'],axis=1)
```

```python
df.isna().sum()
```

```python
df['bmi']=df['bmi'].fillna(df['bmi'].mean())
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
encode=LabelEncoder()
```

```python
l=['gender','ever_married','work_type','Residence_type','smoking_status']
for i in l:
    df[i]=encode.fit_transform(df[i])
```

```python
df.dtypes
```

```python
x=df.drop(['stroke'],axis=1)
```

```python
y=df['stroke']
```

```python
sc=MinMaxScaler()
x=sc.fit_transform(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=1)
```

```python
model=GaussianNB()
```

```python
model.fit(x_train,y_train)
```

```python
y_pre_test=model.predict(x_test)
y_pre_train=model.predict(x_train)
```

```python
print(accuracy_score(y_test,y_pre_test))
print(accuracy_score(y_train,y_pre_train))
```

```python
print(classification_report(y_test,y_pre_test))
```
