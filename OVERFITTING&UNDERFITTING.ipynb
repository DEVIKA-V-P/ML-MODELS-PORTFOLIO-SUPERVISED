```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('5.Melbourne_housing_FULL.csv')
```

```python
cols=['Suburb','Rooms', 'Type', 'Price', 'Method', 'SellerG',
        'Distance', 'Bedroom2', 'Bathroom', 'Car',
       'Landsize', 'BuildingArea', 'CouncilArea',
       'Regionname', 'Propertycount']
```

```python
data=df[cols]
data
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
cols_zero=['Distance','CouncilArea','Regionname','Propertycount']
```

```python
data[cols_zero]=data[cols_zero].fillna(0)
```

```python
data[cols_zero].isna().sum()
```

```python
cols_mean=['Bedroom2','Bathroom','Car','Landsize','BuildingArea']
for i in cols_mean:
    data[i]=data[i].fillna(data[i].mean())
```

```python
data.isna().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data.dtypes
```

```python
data=pd.get_dummies(data,drop_first=True).astype('int')
```

```python
data
```

```python
X=data.drop('Price',axis=1)
y=data.Price
```

```python
from sklearn.preprocessing import MinMaxScaler
m=MinMaxScaler()
X=m.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=1)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred_test=model.predict(X_test)
y_pred_train=model.predict(X_train)
```

```python
from sklearn.metrics import r2_score
print(r2_score(y_test,y_pred_test)*100)
```

```python
from sklearn.metrics import r2_score
print(r2_score(y_train,y_pred_train)*100)
```
