```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv("16.cancer_data.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
x=scaler.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=1)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
# randomforest classifier

# Adaboost classifier
# XGboost classifer
# gredientboost classifier
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import BaggingClassifier
k=KNeighborsClassifier()
b = BaggingClassifier(
    estimator=k,
    n_estimators=200)
```

```python
# randomforest classifier (bugging)
from sklearn.ensemble import RandomForestClassifier
```

```python
rf=RandomForestClassifier(random_state=1,n_estimators=200,criterion='entropy')
rf.fit(x_train,y_train)
```

```python
y_pred_test=rf.predict(x_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
 # adaboost
from sklearn.ensemble import AdaBoostClassifier
ad=AdaBoostClassifier(random_state=1,n_estimators=200)
ad.fit(x_train,y_train)
```

```python
y_ad_test=ad.predict(x_test)
```

```python
print(classification_report(y_test,y_ad_test))
```

```python
# gredientboost classifier
from sklearn.ensemble import GradientBoostingClassifier
gd=GradientBoostingClassifier(random_state=1,n_estimators=200)
gd.fit(x_train,y_train)
y_gd_test=gd.predict(x_test)
```

```python
print(classification_report(y_test,y_gd_test))
```

```python
# xgboosrt classifier
#pip install xgboost
from xgboost import XGBClassifier
#help(XGBClassifier)
xg=XGBClassifier(random_state=1,n_estimator=200)
```

```python
xg.fit(x_train,y_train)
```

```python
y_xg_test=xg.predict(x_test)
print(classification_report(y_test,y_xg_test))
```
