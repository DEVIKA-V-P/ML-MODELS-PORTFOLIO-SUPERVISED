```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as np
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("12.breast-cancer.csv")
df.head()
```

```python
df.info()
```

```python
x=df.iloc[:,2:]
x
```

```python
y=df.iloc[:,1]
y
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.metrics import accuracy_score
```

```python
scaler=MinMaxScaler()
```

```python
x=scaler.fit_transform(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=0)
```

```python
Knn=KNeighborsClassifier(n_neighbors=3)
```

```python
Knn.fit(x_train,y_train)
```

```python
y_pred_test=Knn.predict(x_test)
```

```python
print(accuracy_score(y_test,y_pred_test))
```

```python
y_pred_train=Knn.predict(x_train)
```

```python
print(accuracy_score(y_train,y_pred_train))
```

```python
#hyperparameter tunning
#gridsearchcv
from sklearn.model_selection  import GridSearchCV
```

```python
parameter = {
    'n_neighbors': [1, 3, 5, 7, 9, 10,13],
    'weights': ['uniform','distance'],
    'algorithm': ['auto','ball_tree','kd_tree','brute']
}
```

```python
t=GridSearchCV(Knn,parameter,cv=5,scoring='accuracy')
```

```python
t.fit(x_train,y_train)
```

```python
print(t.best_params_)
```

```python
Kn=KNeighborsClassifier(n_neighbors=7,algorithm='auto',weights= 'uniform')
```

```python
Kn.fit(x_train,y_train)
y_pred_test1=Kn.predict(x_test)
y_pred_train1=Kn.predict(x_train)
print(accuracy_score(y_test,y_pred_test1))
print(accuracy_score(y_train,y_pred_train1))
```

```python
# randomizedsearchcv
from sklearn.model_selection  import RandomizedSearchCV
```

```python
p = {
    'n_neighbors': [1, 3, 5, 7, 9, 10,13],
    'weights': ['uniform','distance'],
    'algorithm': ['auto','ball_tree','kd_tree','brute']
}
```

```python
r=RandomizedSearchCV(Knn,p,cv=5,scoring='accuracy')
r.fit(x_train,y_train)
print(r.best_params_)
```

```python
Kn1=KNeighborsClassifier(n_neighbors=7,algorithm='kd_tree',weights= 'distance')
Kn1.fit(x_train,y_train)
y_pred_test2=Kn1.predict(x_test)
y_pred_train2=Kn1.predict(x_train)
print(accuracy_score(y_test,y_pred_test2))
print(accuracy_score(y_train,y_pred_train2))
```
